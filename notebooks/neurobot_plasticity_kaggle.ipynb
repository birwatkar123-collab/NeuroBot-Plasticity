# NeuroBot-Plasticity

Neuroplasticity-inspired robot task learning with curriculum learning, replay consolidation, and continual-learning evaluation.

Research question: can replay-based consolidation reduce forgetting when a robot policy learns one task and then adapts to another?

## Neuroplasticity Mapping

| Neuroscience principle | ML implementation |
| --- | --- |
| Memory consolidation | replay buffer |
| Gradual skill acquisition | curriculum learning |
| Adaptation to new tasks | sequential Task A then Task B training |
| Stability-plasticity tradeoff | old-task forgetting vs new-task error |

## Interpretation

The replay sweep shows a stability-plasticity tradeoff. Increasing replay reduces forgetting of Task A, but it also increases error on Task B. This is useful because robot learning should not only optimize fast adaptation; it should also preserve useful old skills.
